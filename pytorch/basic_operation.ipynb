# 基础算术运算 (Element-wise)

```python
import torch
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])

# 逐元素相加
print("== a+b ==")
print(a + b)
print(torch.add(a, b))

# 逐元素相乘
print("== a*b ==")
print(a * b)
print(torch.mul(a, b))

print("== a-b ==")
print(a - b)
print(torch.sub(a, b))

print("== a/b ==")
print(a / b)
print(torch.div(a, b))

print("== a**b ==")
print(b ** a)
print(torch.pow(b, a))
```
