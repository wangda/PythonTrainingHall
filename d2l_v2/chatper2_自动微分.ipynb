## 1. 简单例子

```python
import torch

x = torch.arange(4.0)
x
```

```python
x.requires_grad_(True)  # 等价于x=torch.arange(4.0,requires_grad=True)
x.grad  # 默认值是None
```

```python
y = 2 * torch.dot(x, x)
y
```

```python
# 通过运行反向传播函数，计算所有变量的梯度
y.backward()

# 打印梯度
x.grad
```

```python
x.grad == 4 * x
```

```python
# 在默认情况下，PyTorch会累积梯度，我们需要清除之前的值
x.grad.zero_()
y = x.sum()
y.backward()
x.grad
```

```python
# 在默认情况下，PyTorch会累积梯度，我们需要清除之前的值
x.grad.zero_()
y = x.sum()
y.backward()
x.grad
```
